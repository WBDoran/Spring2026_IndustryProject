# Cluster Visualization Notebook: PCA + UMAP

This notebook visualizes the saved V11 lifecycle clusters using PCA and UMAP.

It does **not** recluster developers. It reads the saved cluster membership table from DuckDB, joins cluster labels back to `dev_profile_final_v4`, and creates visualization samples for PCA/UMAP.

This version is aligned with the latest V11 methodology:
- `active`, `cooling`, and `at_risk` are modeled HDBSCAN strata.
- `dormant` is lightweight rule-based segmentation and is excluded from the main PCA/UMAP by default.
- `unactivated` is a pseudo-group and is excluded from the main PCA/UMAP by default.
- Column-name conflicts from merges are handled safely, so `stratum`, `cluster_key`, and `hdbscan_cluster` are normalized before sampling.

Recommended use:
1. Run the V11 clustering notebook first.
2. Confirm `dev_lifecycle_cluster_membership_v11_final` exists.
3. Run this visualization notebook.


```python
# ============================================================
# 0. Imports and setup
# ============================================================

import gc
import warnings
from pathlib import Path

import duckdb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler

warnings.filterwarnings("ignore", category=FutureWarning)

try:
    import umap
except ImportError:
    try:
        import umap.umap_ as umap
    except ImportError as exc:
        raise ImportError(
            "UMAP is not installed. Install it once with: pip install umap-learn"
        ) from exc

DB_PATH = globals().get("DB_PATH", "developer_project.duckdb")
PROFILE_TABLE = globals().get("PROFILE_TABLE", "dev_profile_final_v4")
ID_COL = globals().get("ID_COL", "developer_id")

OUTPUT_TABLE_PREFIX = "dev_lifecycle_cluster_membership_v11"
FINAL_TABLE = f"{OUTPUT_TABLE_PREFIX}_final"
PROFILE_SUMMARY_TABLE = f"{OUTPUT_TABLE_PREFIX}_profile_summary"
RUN_STATS_TABLE = "dev_lifecycle_cluster_run_stats_v11"

OUTPUT_DIR = Path("cluster_visualizations_v11")
OUTPUT_DIR.mkdir(exist_ok=True)

RANDOM_STATE = 42
```

```python
# ============================================================
# 1. Connect to DuckDB and check whether clusters are saved
# ============================================================

con = duckdb.connect(DB_PATH)


def duckdb_table_exists(con, table_name):
    return con.execute(f"""
        SELECT COUNT(*)
        FROM information_schema.tables
        WHERE table_name = '{table_name}'
    """).fetchone()[0] > 0

required_tables = [PROFILE_TABLE, FINAL_TABLE]
optional_tables = [PROFILE_SUMMARY_TABLE]

status_rows = []
for table in required_tables + optional_tables:
    exists = duckdb_table_exists(con, table)
    n_rows = None
    if exists:
        n_rows = con.execute(f"SELECT COUNT(*) FROM {table}").fetchone()[0]
    status_rows.append({"table_name": table, "exists": exists, "n_rows": n_rows})

status_df = pd.DataFrame(status_rows)
display(status_df)

missing_required = [r["table_name"] for r in status_rows[:len(required_tables)] if not r["exists"]]
if missing_required:
    raise ValueError(
        "Missing required saved cluster/profile tables: " + ", ".join(missing_required) +
        ". Run the V11 clustering notebook through the final-table rebuild step first."
    )

print("Clusters are saved and ready for visualization.")
```

```python
# ============================================================
# 2. Load cluster membership and inspect counts
# ============================================================

membership_cols = con.execute(f"DESCRIBE {FINAL_TABLE}").df()["column_name"].tolist()
print("Final membership columns:", membership_cols)

required_membership_cols = [ID_COL, "stratum", "cluster_key", "hdbscan_cluster"]
missing_required_membership = [c for c in required_membership_cols if c not in membership_cols]
if missing_required_membership:
    raise ValueError(
        f"Missing required columns from {FINAL_TABLE}: {missing_required_membership}"
    )

optional_membership_cols = [
    "cluster_probability",
    "outlier_score",
    "adoption_direction",
    "proposed_business_label",
]
select_membership_cols = required_membership_cols + [
    c for c in optional_membership_cols if c in membership_cols
]

membership_df = con.execute(f"""
    SELECT {", ".join(select_membership_cols)}
    FROM {FINAL_TABLE}
""").df()

print("Membership rows:", len(membership_df))

cluster_counts = (
    membership_df
    .groupby(["stratum", "cluster_key"], dropna=False)
    .size()
    .reset_index(name="n_developers")
    .sort_values(["stratum", "n_developers"], ascending=[True, False])
)

display(cluster_counts)
```

```python
# ============================================================
# 3. Configuration for visualization sampling
# ============================================================

# PCA/UMAP can be expensive and unreadable on all 9M+ developers.
# The clustering itself is full-population; visualization uses a stratified sample.

MODELED_STRATA = ["active", "cooling", "at_risk"]
INCLUDE_NOISE = False
INCLUDE_DORMANT_AND_UNACTIVATED = False

# Recommended range: 3,000 to 10,000 per cluster.
# Increase if your machine can handle it.
VIZ_SAMPLE_PER_CLUSTER = 5000
MAX_TOTAL_VIZ_ROWS = 150000

# Latest V11 visualization features.
# These are intentionally shared behavioral interpretation features across active/cooling/at_risk.
VIZ_FEATURES = [
    "log_activity_count_0_30d",
    "log_activity_count_30_90d",
    "log_activity_count_90_180d",
    "unique_activity_types_0_30d",
    "unique_modalities_0_30d",
    "developer_effort_score",
    "weighted_recent_confidence_effort",
    "recent_build_flag",
    "log_build_count_0_30d",
    "build_share_lifetime",
    "high_effort_share_lifetime",
    "log_clipped_lifetime_activity_count_p99",
    "persona_entropy",
]

print("Modeled strata for PCA/UMAP:", MODELED_STRATA)
print("Include noise:", INCLUDE_NOISE)
print("Include dormant/unactivated:", INCLUDE_DORMANT_AND_UNACTIVATED)
print("Sample per cluster:", VIZ_SAMPLE_PER_CLUSTER)
print("Max total visualization rows:", MAX_TOTAL_VIZ_ROWS)
```

```python
# ============================================================
# 4. Load visualization features from profile table
# ============================================================

profile_cols = con.execute(f"DESCRIBE {PROFILE_TABLE}").df()["column_name"].tolist()
existing_viz_features = [c for c in VIZ_FEATURES if c in profile_cols]
missing_viz_features = [c for c in VIZ_FEATURES if c not in profile_cols]

print("Existing visualization features:", len(existing_viz_features))
print(existing_viz_features)

if missing_viz_features:
    print("Missing visualization features. They will be skipped:")
    for col in missing_viz_features:
        print("-", col)

if not existing_viz_features:
    raise ValueError("No visualization features were found in the profile table.")

select_cols = [ID_COL] + existing_viz_features
profile_feature_df = con.execute(f"""
    SELECT {", ".join(select_cols)}
    FROM {PROFILE_TABLE}
""").df()

print("Profile feature rows:", len(profile_feature_df))
```

```python
# ============================================================
# 5. Build visualization dataframe safely
# ============================================================

viz_df = membership_df.merge(
    profile_feature_df,
    on=ID_COL,
    how="left",
    suffixes=("_membership", "_profile"),
)

print("viz_df columns after merge:")
print(viz_df.columns.tolist())

# Normalize key columns in case any merge or prior notebook produced suffixes.
def ensure_column(df, target_col, candidates):
    if target_col in df.columns:
        return df
    for cand in candidates:
        if cand in df.columns:
            df[target_col] = df[cand]
            return df
    raise KeyError(
        f"Could not find `{target_col}`. Tried: {[target_col] + candidates}. "
        f"Available columns: {df.columns.tolist()}"
    )

viz_df = ensure_column(
    viz_df,
    "stratum",
    ["stratum_membership", "stratum_profile", "stratum_x", "stratum_y", "dormancy_segment", "lifecycle_stratum"],
)

viz_df = ensure_column(
    viz_df,
    "cluster_key",
    ["cluster_key_membership", "cluster_key_profile", "cluster_key_x", "cluster_key_y"],
)

if "hdbscan_cluster" not in viz_df.columns:
    for cand in ["hdbscan_cluster_membership", "hdbscan_cluster_profile", "hdbscan_cluster_x", "hdbscan_cluster_y"]:
        if cand in viz_df.columns:
            viz_df["hdbscan_cluster"] = viz_df[cand]
            break
if "hdbscan_cluster" not in viz_df.columns:
    viz_df["hdbscan_cluster"] = np.nan

# Coerce string columns to consistent text.
viz_df["stratum"] = viz_df["stratum"].astype(str).str.lower().str.strip()
viz_df["cluster_key"] = viz_df["cluster_key"].astype(str)

if INCLUDE_DORMANT_AND_UNACTIVATED:
    keep_strata = ["active", "cooling", "at_risk", "dormant", "unactivated"]
else:
    keep_strata = MODELED_STRATA

viz_df = viz_df[viz_df["stratum"].isin(keep_strata)].copy()

if not INCLUDE_NOISE:
    viz_df = viz_df[
        ~viz_df["cluster_key"].astype(str).str.contains("noise", na=False)
    ].copy()

# Drop rows with no usable feature values.
viz_df[existing_viz_features] = viz_df[existing_viz_features].replace([np.inf, -np.inf], np.nan)
viz_df = viz_df[viz_df[existing_viz_features].notna().any(axis=1)].copy()

print("Rows available for visualization before sampling:", len(viz_df))
display(
    viz_df.groupby(["stratum", "cluster_key"], dropna=False)
    .size()
    .reset_index(name="n_available")
    .sort_values(["stratum", "n_available"], ascending=[True, False])
)
```

```python
# ============================================================
# 6. Stratified sample by cluster
# Robust version: avoids pandas groupby/apply column-loss issues
# ============================================================

# Re-normalize key columns here too, in case cells were run out of order.
def recover_column(df, target_col, candidates):
    if target_col in df.columns:
        return df
    for cand in candidates:
        if cand in df.columns:
            df[target_col] = df[cand]
            return df
    raise KeyError(
        f"Missing `{target_col}`. Tried candidates: {candidates}. "
        f"Available columns: {df.columns.tolist()}"
    )

viz_df = recover_column(
    viz_df,
    "stratum",
    [
        "stratum_membership",
        "stratum_profile",
        "stratum_x",
        "stratum_y",
        "dormancy_segment",
        "lifecycle_stratum",
    ],
)

viz_df = recover_column(
    viz_df,
    "cluster_key",
    [
        "cluster_key_membership",
        "cluster_key_profile",
        "cluster_key_x",
        "cluster_key_y",
    ],
)

if "hdbscan_cluster" not in viz_df.columns:
    for cand in [
        "hdbscan_cluster_membership",
        "hdbscan_cluster_profile",
        "hdbscan_cluster_x",
        "hdbscan_cluster_y",
    ]:
        if cand in viz_df.columns:
            viz_df["hdbscan_cluster"] = viz_df[cand]
            break

if "hdbscan_cluster" not in viz_df.columns:
    viz_df["hdbscan_cluster"] = np.nan

viz_df["stratum"] = viz_df["stratum"].astype(str).str.lower().str.strip()
viz_df["cluster_key"] = viz_df["cluster_key"].astype(str)

print("viz_df columns before sampling:")
print(viz_df.columns.tolist())
print("Rows before sampling:", len(viz_df))

required_for_sampling = ["stratum", "cluster_key"]
missing_for_sampling = [c for c in required_for_sampling if c not in viz_df.columns]
if missing_for_sampling:
    raise KeyError(f"Missing required sampling columns: {missing_for_sampling}")

if len(viz_df) == 0:
    raise ValueError("No rows available for visualization after filtering.")

# Avoid groupby.apply because some pandas versions / notebook states can drop columns unexpectedly.
sampled_parts = []
for (stratum_value, cluster_value), group in viz_df.groupby(
    ["stratum", "cluster_key"],
    dropna=False,
    sort=False,
):
    n_take = min(len(group), VIZ_SAMPLE_PER_CLUSTER)
    sampled_parts.append(group.sample(n=n_take, random_state=RANDOM_STATE))

viz_sample = pd.concat(sampled_parts, ignore_index=True) if sampled_parts else pd.DataFrame()

# Optional global cap. This keeps the notebook from becoming too slow.
if len(viz_sample) > MAX_TOTAL_VIZ_ROWS:
    viz_sample = (
        viz_sample
        .sample(n=MAX_TOTAL_VIZ_ROWS, random_state=RANDOM_STATE)
        .reset_index(drop=True)
    )

# Final safety check after sampling/capping.
viz_sample = recover_column(
    viz_sample,
    "stratum",
    [
        "stratum_membership",
        "stratum_profile",
        "stratum_x",
        "stratum_y",
        "dormancy_segment",
        "lifecycle_stratum",
    ],
)

viz_sample = recover_column(
    viz_sample,
    "cluster_key",
    [
        "cluster_key_membership",
        "cluster_key_profile",
        "cluster_key_x",
        "cluster_key_y",
    ],
)

if "hdbscan_cluster" not in viz_sample.columns:
    viz_sample["hdbscan_cluster"] = np.nan

print("Visualization sample rows:", len(viz_sample))
print("viz_sample columns after sampling:")
print(viz_sample.columns.tolist())

display(
    viz_sample
    .groupby(["stratum", "cluster_key"], dropna=False)
    .size()
    .reset_index(name="n_sampled")
    .sort_values(["stratum", "n_sampled"], ascending=[True, False])
)

sample_path = OUTPUT_DIR / "cluster_visualization_sample_v11.csv"
export_cols = [ID_COL, "stratum", "cluster_key", "hdbscan_cluster"] + existing_viz_features
export_cols = [c for c in export_cols if c in viz_sample.columns]
viz_sample[export_cols].to_csv(sample_path, index=False)
print("Saved visualization sample:", sample_path)

```

```python
# ============================================================
# 7. Preprocess feature matrix
# ============================================================

X_raw = viz_sample[existing_viz_features].replace([np.inf, -np.inf], np.nan)

prep = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", RobustScaler()),
])

X = prep.fit_transform(X_raw).astype("float32")

# Drop zero-variance columns after preprocessing.
keep_mask = np.nanvar(X, axis=0) > 1e-12
kept_features = [f for f, keep in zip(existing_viz_features, keep_mask) if keep]
dropped_features = [f for f, keep in zip(existing_viz_features, keep_mask) if not keep]
X = X[:, keep_mask]

if X.shape[1] == 0:
    raise ValueError("All visualization features became zero-variance after preprocessing.")

print("Final feature matrix shape:", X.shape)
print("Kept features:", kept_features)
if dropped_features:
    print("Dropped zero-variance features:", dropped_features)
```

```python
# ============================================================
# 8. PCA projection
# ============================================================

pca = PCA(n_components=2, random_state=RANDOM_STATE)
pca_xy = pca.fit_transform(X)

viz_sample["pca_1"] = pca_xy[:, 0]
viz_sample["pca_2"] = pca_xy[:, 1]

print("PCA explained variance ratio:", pca.explained_variance_ratio_)
print("Total explained variance:", pca.explained_variance_ratio_.sum())

pca_path = OUTPUT_DIR / "cluster_pca_coordinates_v11.csv"
viz_sample[[ID_COL, "stratum", "cluster_key", "pca_1", "pca_2"]].to_csv(pca_path, index=False)
print("Saved PCA coordinates:", pca_path)
```

```python
# ============================================================
# 9A. PCA colored by cluster
# ============================================================

plt.figure(figsize=(11, 7))

for cluster_key, g in viz_sample.groupby("cluster_key"):
    plt.scatter(
        g["pca_1"],
        g["pca_2"],
        s=7,
        alpha=0.45,
        label=cluster_key,
    )

plt.title("PCA projection of V11 developer clusters")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.legend(markerscale=2, bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()

pca_cluster_png = OUTPUT_DIR / "pca_by_cluster_v11.png"
plt.savefig(pca_cluster_png, dpi=160, bbox_inches="tight")
plt.show()

print("Saved:", pca_cluster_png)
```

```python
# ============================================================
# 9B. PCA colored by lifecycle stratum
# ============================================================

plt.figure(figsize=(9, 6))

for stratum, g in viz_sample.groupby("stratum"):
    plt.scatter(
        g["pca_1"],
        g["pca_2"],
        s=7,
        alpha=0.45,
        label=stratum,
    )

plt.title("PCA projection by lifecycle stratum")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.legend(markerscale=2)
plt.tight_layout()

pca_stratum_png = OUTPUT_DIR / "pca_by_stratum_v11.png"
plt.savefig(pca_stratum_png, dpi=160, bbox_inches="tight")
plt.show()

print("Saved:", pca_stratum_png)
```

```python
# ============================================================
# 10. UMAP projection
# ============================================================

# UMAP is nonlinear and better for local cluster structure than PCA.
# Runtime depends heavily on the number of sampled rows.

umap_model = umap.UMAP(
    n_components=2,
    n_neighbors=30,
    min_dist=0.10,
    metric="euclidean",
    random_state=RANDOM_STATE,
    low_memory=True,
)

umap_xy = umap_model.fit_transform(X)

viz_sample["umap_1"] = umap_xy[:, 0]
viz_sample["umap_2"] = umap_xy[:, 1]

umap_path = OUTPUT_DIR / "cluster_umap_coordinates_v11.csv"
viz_sample[[ID_COL, "stratum", "cluster_key", "umap_1", "umap_2"]].to_csv(umap_path, index=False)
print("Saved UMAP coordinates:", umap_path)
```

```python
# ============================================================
# 11A. UMAP colored by cluster
# ============================================================

plt.figure(figsize=(11, 7))

for cluster_key, g in viz_sample.groupby("cluster_key"):
    plt.scatter(
        g["umap_1"],
        g["umap_2"],
        s=7,
        alpha=0.45,
        label=cluster_key,
    )

plt.title("UMAP projection of V11 developer clusters")
plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.legend(markerscale=2, bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()

umap_cluster_png = OUTPUT_DIR / "umap_by_cluster_v11.png"
plt.savefig(umap_cluster_png, dpi=160, bbox_inches="tight")
plt.show()

print("Saved:", umap_cluster_png)
```

```python
# ============================================================
# 11B. UMAP colored by lifecycle stratum
# ============================================================

plt.figure(figsize=(9, 6))

for stratum, g in viz_sample.groupby("stratum"):
    plt.scatter(
        g["umap_1"],
        g["umap_2"],
        s=7,
        alpha=0.45,
        label=stratum,
    )

plt.title("UMAP projection by lifecycle stratum")
plt.xlabel("UMAP 1")
plt.ylabel("UMAP 2")
plt.legend(markerscale=2)
plt.tight_layout()

umap_stratum_png = OUTPUT_DIR / "umap_by_stratum_v11.png"
plt.savefig(umap_stratum_png, dpi=160, bbox_inches="tight")
plt.show()

print("Saved:", umap_stratum_png)
```

```python
# ============================================================
# 12. Cluster size chart from full data
# ============================================================

cluster_size_plot = (
    membership_df
    .groupby(["stratum", "cluster_key"])
    .size()
    .reset_index(name="n_developers")
)

cluster_size_plot = cluster_size_plot[
    ~cluster_size_plot["cluster_key"].str.contains("noise", na=False)
].copy()

cluster_size_plot = cluster_size_plot.sort_values(
    ["stratum", "n_developers"],
    ascending=[True, False]
)

cluster_size_plot["label"] = (
    cluster_size_plot["stratum"] + ": " + cluster_size_plot["cluster_key"]
)

plt.figure(figsize=(11, max(5, 0.32 * len(cluster_size_plot))))
plt.barh(cluster_size_plot["label"], cluster_size_plot["n_developers"])
plt.title("Full-data cluster sizes by lifecycle stratum")
plt.xlabel("Developers")
plt.ylabel("Cluster")
plt.gca().invert_yaxis()
plt.tight_layout()

cluster_size_png = OUTPUT_DIR / "full_data_cluster_sizes_v11.png"
plt.savefig(cluster_size_png, dpi=160, bbox_inches="tight")
plt.show()

print("Saved:", cluster_size_png)
```

```python
# ============================================================
# 13. Optional: cluster centroids projected into PCA space
# ============================================================

centroids = (
    viz_sample
    .groupby(["stratum", "cluster_key"], as_index=False)
    .agg(
        pca_1=("pca_1", "mean"),
        pca_2=("pca_2", "mean"),
        umap_1=("umap_1", "mean"),
        umap_2=("umap_2", "mean"),
        n_sampled=(ID_COL, "count"),
    )
)

display(centroids)

centroid_path = OUTPUT_DIR / "cluster_visualization_centroids_v11.csv"
centroids.to_csv(centroid_path, index=False)
print("Saved centroids:", centroid_path)
```

```python
# ============================================================
# 14. Interpretation notes
# ============================================================

print("""
How to read these visuals:

1. PCA shows broad linear separation in the feature space.
2. UMAP shows local/nonlinear separation and is usually better for presentation.
3. Cluster size charts are full-data charts, not sampled charts.
4. PCA/UMAP are diagnostics, not proof that the clusters are correct.
5. The production output remains dev_lifecycle_cluster_membership_v11_final.
6. The main PCA/UMAP intentionally focuses on active, cooling, and at_risk because those are the modeled HDBSCAN strata.
7. Dormant and unactivated are excluded by default because they are rule-based or pseudo-groups, not behavioral HDBSCAN clusters.
""")

con.close()
print("DuckDB connection closed.")
```

```python

# ============================================================
# Presentation visualization setup for all clustering models
# ============================================================

import duckdb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

DB_PATH = globals().get("DB_PATH", "developer_project.duckdb")
PRESENTATION_DIR = Path("presentation_cluster_visuals")
PRESENTATION_DIR.mkdir(exist_ok=True)

con = duckdb.connect(DB_PATH)

NVIDIA_GREEN = "#76B900"
NVIDIA_DARK = "#1A1A1A"
NVIDIA_GRAY = "#666666"
NVIDIA_LIGHT_GRAY = "#D9D9D9"

STRATUM_PALETTE = {
    "active": NVIDIA_GREEN,
    "cooling": "#F4B400",
    "at_risk": "#DB4437",
    "dormant": "#5F6368",
    "unactivated": "#B0BEC5",
    "unknown": "#9E9E9E",
}

MODEL_PALETTE = {
    "HDBSCAN": NVIDIA_GREEN,
    "GMM": "#2E7D32",
    "Weekly GMM": "#43A047",
    "HMM": "#1E88E5",
    "LightGBM": "#F4B400",
    "XGBoost": "#DB4437",
    "Topic Modeling": "#6A1B9A",
}

ACTIVE_CLUSTER_COLORS = ["#76B900", "#5AA700", "#3F8F00", "#98D62E", "#4F6F1A", "#B7E46A", "#2E7D32", "#A5D6A7"]
COOLING_CLUSTER_COLORS = ["#F4B400", "#E09F00", "#C98C00", "#FFCA28", "#D6A700", "#A67C00", "#FFE082", "#8D6E00"]
ATRISK_CLUSTER_COLORS = ["#DB4437", "#C53929", "#B71C1C", "#EF5350", "#E57373", "#8E0000", "#FFCDD2", "#A52714"]
GENERAL_CLUSTER_COLORS = [NVIDIA_GREEN, "#1E88E5", "#F4B400", "#DB4437", "#6A1B9A", "#00897B", "#5F6368", "#8D6E63", "#3949AB", "#C0CA33"]

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.edgecolor": "#333333",
    "axes.labelcolor": "#222222",
    "xtick.color": "#222222",
    "ytick.color": "#222222",
    "font.size": 11,
    "axes.titleweight": "bold",
})

def table_exists(table_name):
    return con.execute(f"""
        SELECT COUNT(*)
        FROM information_schema.tables
        WHERE table_name = '{table_name}'
    """).fetchone()[0] > 0

def save_current_fig(name):
    path = PRESENTATION_DIR / f"{name}.png"
    plt.savefig(path, dpi=180, bbox_inches="tight")
    print("Saved:", path)

def clean_label(x):
    return str(x).replace("_", " ").title()

def add_value_labels(ax, fmt=",.0f", padding=3):
    for p in ax.patches:
        width = p.get_width()
        ax.text(
            width + padding,
            p.get_y() + p.get_height() / 2,
            format(width, fmt),
            va="center",
            fontsize=9,
            color=NVIDIA_DARK,
        )

print("Connected to", DB_PATH)
print("Outputs will save to", PRESENTATION_DIR.resolve())
```

```python

# ============================================================
# HDBSCAN Visual 1: Lifecycle funnel / population overview
# ============================================================

HDBSCAN_FINAL = "dev_lifecycle_cluster_membership_v11_final"

if table_exists(HDBSCAN_FINAL):
    hdbscan_counts = con.execute(f"""
        SELECT stratum, COUNT(*) AS n_developers
        FROM {HDBSCAN_FINAL}
        GROUP BY stratum
        ORDER BY n_developers DESC
    """).df()

    hdbscan_counts["color"] = hdbscan_counts["stratum"].map(STRATUM_PALETTE).fillna(NVIDIA_GRAY)
    hdbscan_counts["label"] = hdbscan_counts["stratum"].map(clean_label)

    fig, ax = plt.subplots(figsize=(11, 6))
    ax.barh(hdbscan_counts["label"], hdbscan_counts["n_developers"], color=hdbscan_counts["color"])
    ax.invert_yaxis()
    ax.set_title("Developer Lifecycle Distribution", fontsize=18)
    ax.set_xlabel("Developers")
    ax.set_ylabel("")
    ax.grid(axis="x", alpha=0.15)
    add_value_labels(ax, fmt=",.0f", padding=max(hdbscan_counts["n_developers"]) * 0.01)
    plt.tight_layout()
    save_current_fig("hdbscan_lifecycle_distribution")
    plt.show()
else:
    print(f"Missing table: {HDBSCAN_FINAL}")
```

```python

# ============================================================
# HDBSCAN Visual 2: Cluster sizes by lifecycle stratum
# ============================================================

if table_exists(HDBSCAN_FINAL):
    hdbscan_cluster_sizes = con.execute(f"""
        SELECT stratum, cluster_key, COUNT(*) AS n_developers
        FROM {HDBSCAN_FINAL}
        GROUP BY stratum, cluster_key
        ORDER BY stratum, n_developers DESC
    """).df()

    hdbscan_cluster_sizes = hdbscan_cluster_sizes[
        ~hdbscan_cluster_sizes["cluster_key"].astype(str).str.contains("noise", case=False, na=False)
    ].copy()
    hdbscan_cluster_sizes["label"] = hdbscan_cluster_sizes["stratum"].map(clean_label) + ": " + hdbscan_cluster_sizes["cluster_key"].astype(str)
    hdbscan_cluster_sizes["color"] = hdbscan_cluster_sizes["stratum"].map(STRATUM_PALETTE).fillna(NVIDIA_GRAY)

    fig, ax = plt.subplots(figsize=(12, max(6, 0.35 * len(hdbscan_cluster_sizes))))
    ax.barh(hdbscan_cluster_sizes["label"], hdbscan_cluster_sizes["n_developers"], color=hdbscan_cluster_sizes["color"])
    ax.invert_yaxis()
    ax.set_title("HDBSCAN Cluster Sizes by Lifecycle Stage", fontsize=18)
    ax.set_xlabel("Developers")
    ax.set_ylabel("")
    ax.grid(axis="x", alpha=0.15)
    plt.tight_layout()
    save_current_fig("hdbscan_cluster_sizes")
    plt.show()
else:
    print(f"Missing table: {HDBSCAN_FINAL}")
```

```python

# ============================================================
# HDBSCAN Visual 3: Cluster feature fingerprint heatmap
# ============================================================

PROFILE_TABLE = "dev_lifecycle_cluster_membership_v11_profile_summary"
HEATMAP_FEATURES = [
    "log_activity_count_0_30d",
    "log_activity_count_30_90d",
    "log_activity_count_90_180d",
    "developer_effort_score",
    "weighted_recent_confidence_effort",
    "build_share_lifetime",
    "high_effort_share_lifetime",
    "log_clipped_lifetime_activity_count_p99",
    "persona_entropy",
]

if table_exists(PROFILE_TABLE):
    profile_cols = con.execute(f"DESCRIBE {PROFILE_TABLE}").df()["column_name"].tolist()
    features = [c for c in HEATMAP_FEATURES if c in profile_cols]

    prof = con.execute(f"SELECT * FROM {PROFILE_TABLE}").df()
    prof = prof[
        prof["stratum"].isin(["active", "cooling", "at_risk"])
        & ~prof["cluster_key"].astype(str).str.contains("noise", case=False, na=False)
    ].copy()
    prof = prof.sort_values(["stratum", "n_developers"], ascending=[True, False])

    if len(prof) and features:
        X = prof[features].replace([np.inf, -np.inf], np.nan).fillna(0.0)
        Z = (X - X.mean(axis=0)) / X.std(axis=0).replace(0, 1)

        fig, ax = plt.subplots(figsize=(13, max(6, 0.4 * len(prof))))
        im = ax.imshow(Z.values, aspect="auto", cmap="YlGn")
        ax.set_yticks(range(len(prof)))
        ax.set_yticklabels(prof["cluster_key"].astype(str).tolist())
        ax.set_xticks(range(len(features)))
        ax.set_xticklabels([clean_label(f) for f in features], rotation=40, ha="right")
        ax.set_title("HDBSCAN Cluster Behavioral Fingerprints", fontsize=18)
        cbar = plt.colorbar(im, ax=ax)
        cbar.set_label("Z-scored Cluster Mean")
        plt.tight_layout()
        save_current_fig("hdbscan_cluster_fingerprint_heatmap")
        plt.show()
    else:
        print("No profile rows/features available for heatmap.")
else:
    print(f"Missing table: {PROFILE_TABLE}")
```

```python

# ============================================================
# HDBSCAN Visual 4: Cleaner UMAP by lifecycle stratum
# Requires viz_sample from earlier notebook cells
# ============================================================

if "viz_sample" in globals() and {"umap_1", "umap_2", "stratum"}.issubset(viz_sample.columns):
    fig, ax = plt.subplots(figsize=(13, 8))

    for stratum, g in viz_sample.groupby("stratum"):
        ax.scatter(
            g["umap_1"], g["umap_2"],
            s=13,
            alpha=0.55,
            color=STRATUM_PALETTE.get(stratum, NVIDIA_GRAY),
            label=clean_label(stratum),
            edgecolors="none",
        )

    ax.set_title("Developer Behavioral Landscape by Lifecycle Stage", fontsize=18)
    ax.set_xlabel("UMAP 1")
    ax.set_ylabel("UMAP 2")
    ax.grid(alpha=0.12)
    ax.legend(frameon=True, title="Lifecycle")
    plt.tight_layout()
    save_current_fig("hdbscan_umap_by_lifecycle_clean")
    plt.show()
else:
    print("viz_sample with UMAP columns not found. Run the PCA/UMAP cells above first.")
```

```python

# ============================================================
# GMM Visuals: Stratified cluster sizes and uncertainty
# ============================================================

GMM_TABLE = "dev_gmm_stratified_clusters_v1"

if table_exists(GMM_TABLE):
    gmm = con.execute(f"SELECT * FROM {GMM_TABLE}").df()

    gmm_sizes = (
        gmm.groupby(["stratum", "gmm_cluster_label"], dropna=False)
        .size()
        .reset_index(name="n_developers")
        .sort_values(["stratum", "n_developers"], ascending=[True, False])
    )
    gmm_sizes["label"] = gmm_sizes["stratum"].map(clean_label) + ": " + gmm_sizes["gmm_cluster_label"].astype(str)
    gmm_sizes["color"] = gmm_sizes["stratum"].map(STRATUM_PALETTE).fillna(MODEL_PALETTE["GMM"])

    fig, ax = plt.subplots(figsize=(12, max(6, 0.35 * len(gmm_sizes))))
    ax.barh(gmm_sizes["label"], gmm_sizes["n_developers"], color=gmm_sizes["color"])
    ax.invert_yaxis()
    ax.set_title("GMM Cluster Sizes by Lifecycle Stage", fontsize=18)
    ax.set_xlabel("Developers")
    ax.set_ylabel("")
    ax.grid(axis="x", alpha=0.15)
    plt.tight_layout()
    save_current_fig("gmm_cluster_sizes")
    plt.show()

    if {"gmm_max_posterior", "gmm_posterior_entropy"}.issubset(gmm.columns):
        gmm_uncertainty = (
            gmm.groupby("stratum", dropna=False)
            .agg(
                avg_max_posterior=("gmm_max_posterior", "mean"),
                avg_entropy=("gmm_posterior_entropy", "mean"),
                n_developers=("developer_id", "count"),
            )
            .reset_index()
        )

        fig, ax = plt.subplots(figsize=(10, 5))
        x = np.arange(len(gmm_uncertainty))
        ax.bar(x - 0.18, gmm_uncertainty["avg_max_posterior"], width=0.36, color=NVIDIA_GREEN, label="Avg Max Posterior")
        ax.bar(x + 0.18, gmm_uncertainty["avg_entropy"], width=0.36, color="#1E88E5", label="Avg Entropy")
        ax.set_xticks(x)
        ax.set_xticklabels(gmm_uncertainty["stratum"].map(clean_label))
        ax.set_title("GMM Assignment Confidence and Ambiguity", fontsize=18)
        ax.set_ylabel("Score")
        ax.legend()
        ax.grid(axis="y", alpha=0.15)
        plt.tight_layout()
        save_current_fig("gmm_confidence_entropy_by_stratum")
        plt.show()
else:
    print(f"Missing table: {GMM_TABLE}")
```

```python

# ============================================================
# Weekly GMM Visual: Weekly state distribution
# ============================================================

WEEKLY_GMM_TABLE = "dev_gmm_weekly_clusters_v1"

if table_exists(WEEKLY_GMM_TABLE):
    weekly_state_counts = con.execute(f"""
        SELECT gmm_weekly_cluster_id, COUNT(*) AS n_weeks
        FROM {WEEKLY_GMM_TABLE}
        GROUP BY gmm_weekly_cluster_id
        ORDER BY gmm_weekly_cluster_id
    """).df()

    colors = [GENERAL_CLUSTER_COLORS[i % len(GENERAL_CLUSTER_COLORS)] for i in range(len(weekly_state_counts))]

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.bar(weekly_state_counts["gmm_weekly_cluster_id"].astype(str), weekly_state_counts["n_weeks"], color=colors)
    ax.set_title("Weekly GMM Behavior-State Distribution", fontsize=18)
    ax.set_xlabel("Weekly GMM State")
    ax.set_ylabel("Developer-Weeks")
    ax.grid(axis="y", alpha=0.15)
    plt.tight_layout()
    save_current_fig("weekly_gmm_state_distribution")
    plt.show()
else:
    print(f"Missing table: {WEEKLY_GMM_TABLE}")
```

```python

# ============================================================
# HMM Visuals: State distribution and transition matrix
# ============================================================

HMM_WEEKLY_TABLE = "dev_hmm_weekly_states_v1"
HMM_TRANSITION_TABLE = "dev_hmm_transition_matrix_v1"

if table_exists(HMM_WEEKLY_TABLE):
    hmm_cols = con.execute(f"DESCRIBE {HMM_WEEKLY_TABLE}").df()["column_name"].tolist()
    state_col = "hmm_state_label" if "hmm_state_label" in hmm_cols else ("hidden_state" if "hidden_state" in hmm_cols else None)

    if state_col:
        hmm_state_counts = con.execute(f"""
            SELECT {state_col} AS hmm_state, COUNT(*) AS n_weeks
            FROM {HMM_WEEKLY_TABLE}
            GROUP BY {state_col}
            ORDER BY n_weeks DESC
        """).df()

        colors = [NVIDIA_GREEN] + [GENERAL_CLUSTER_COLORS[i % len(GENERAL_CLUSTER_COLORS)] for i in range(1, len(hmm_state_counts))]

        fig, ax = plt.subplots(figsize=(10, 5))
        ax.barh(hmm_state_counts["hmm_state"].astype(str), hmm_state_counts["n_weeks"], color=colors[:len(hmm_state_counts)])
        ax.invert_yaxis()
        ax.set_title("HMM Hidden Journey-State Distribution", fontsize=18)
        ax.set_xlabel("Developer-Weeks")
        ax.set_ylabel("HMM State")
        ax.grid(axis="x", alpha=0.15)
        plt.tight_layout()
        save_current_fig("hmm_state_distribution")
        plt.show()
    else:
        print("Could not find an HMM state column in", HMM_WEEKLY_TABLE)
else:
    print(f"Missing table: {HMM_WEEKLY_TABLE}")

if table_exists(HMM_TRANSITION_TABLE):
    trans = con.execute(f"SELECT * FROM {HMM_TRANSITION_TABLE}").df()
    print("Transition matrix columns:", trans.columns.tolist())

    # Try common schemas first.
    from_candidates = ["from_state_label", "from_state", "current_state", "state_from", "source_state"]
    to_candidates = ["to_state_label", "to_state", "next_state", "state_to", "target_state"]
    prob_candidates = ["transition_probability", "probability", "prob", "transition_prob", "p"]

    from_col = next((c for c in from_candidates if c in trans.columns), None)
    to_col = next((c for c in to_candidates if c in trans.columns), None)
    prob_col = next((c for c in prob_candidates if c in trans.columns), None)

    if from_col and to_col and prob_col:
        mat = trans.pivot_table(index=from_col, columns=to_col, values=prob_col, fill_value=0)
        fig, ax = plt.subplots(figsize=(9, 7))
        im = ax.imshow(mat.values, cmap="YlGn", vmin=0, vmax=max(0.01, mat.values.max()))
        ax.set_xticks(range(mat.shape[1]))
        ax.set_xticklabels(mat.columns.astype(str), rotation=35, ha="right")
        ax.set_yticks(range(mat.shape[0]))
        ax.set_yticklabels(mat.index.astype(str))
        ax.set_title("HMM Journey Transition Matrix", fontsize=18)
        cbar = plt.colorbar(im, ax=ax)
        cbar.set_label("Transition Probability")

        for i in range(mat.shape[0]):
            for j in range(mat.shape[1]):
                val = mat.values[i, j]
                if val >= 0.05:
                    ax.text(j, i, f"{val:.2f}", ha="center", va="center", fontsize=8, color=NVIDIA_DARK)

        plt.tight_layout()
        save_current_fig("hmm_transition_matrix")
        plt.show()
    else:
        print("Could not infer transition-matrix columns. Please inspect printed columns above.")
else:
    print(f"Missing table: {HMM_TRANSITION_TABLE}")
```

```python
# ============================================================
# Supervised Visuals: LightGBM/XGBoost cluster prediction outputs
# ============================================================

SUP_FINAL = "dev_supervised_cluster_membership_v1_final"
SUP_STATS = "dev_supervised_cluster_run_stats_v1"

if table_exists(SUP_FINAL):
    sup = con.execute(f"SELECT * FROM {SUP_FINAL}").df()

    # Normalize column names across notebook versions
    model_col = (
        "source_model" if "source_model" in sup.columns
        else "supervised_model" if "supervised_model" in sup.columns
        else "model_type" if "model_type" in sup.columns
        else None
    )

    label_col = (
        "supervised_cluster_label" if "supervised_cluster_label" in sup.columns
        else "predicted_cluster_label" if "predicted_cluster_label" in sup.columns
        else "predicted_cluster_key" if "predicted_cluster_key" in sup.columns
        else "cluster_key" if "cluster_key" in sup.columns
        else None
    )

    if model_col is None:
        print("No model column found. Available columns:", sup.columns.tolist())
    if label_col is None:
        print("No cluster label column found. Available columns:", sup.columns.tolist())

    if model_col and label_col:
        sup_sizes = (
            sup.groupby([model_col, "stratum", label_col], dropna=False)
            .size()
            .reset_index(name="n_developers")
            .sort_values([model_col, "stratum", "n_developers"], ascending=[True, True, False])
        )

        top_sup = sup_sizes.groupby(model_col).head(12).copy()
        top_sup["plot_label"] = (
            top_sup[model_col].astype(str)
            + " | "
            + top_sup["stratum"].astype(str)
            + ": "
            + top_sup[label_col].astype(str)
        )

        top_sup["color"] = (
            top_sup[model_col]
            .astype(str)
            .str.lower()
            .map({
                "lgbm": MODEL_PALETTE["LightGBM"],
                "lightgbm": MODEL_PALETTE["LightGBM"],
                "xgb": MODEL_PALETTE["XGBoost"],
                "xgboost": MODEL_PALETTE["XGBoost"],
            })
            .fillna(NVIDIA_GREEN)
        )

        fig, ax = plt.subplots(figsize=(12, max(6, 0.35 * len(top_sup))))
        ax.barh(top_sup["plot_label"], top_sup["n_developers"], color=top_sup["color"])
        ax.invert_yaxis()
        ax.set_title("Supervised Cluster Assignments by Model", fontsize=18)
        ax.set_xlabel("Developers")
        ax.set_ylabel("")
        ax.grid(axis="x", alpha=0.15)
        plt.tight_layout()
        save_current_fig("supervised_cluster_assignments")
        plt.show()

    conf_col = "prediction_confidence" if "prediction_confidence" in sup.columns else None
    ent_col = "prediction_entropy" if "prediction_entropy" in sup.columns else None

    if model_col and conf_col:
        agg_dict = {
            "avg_confidence": (conf_col, "mean"),
            "n_developers": ("developer_id", "count"),
        }

        if ent_col:
            agg_dict["avg_entropy"] = (ent_col, "mean")

        sup_conf = (
            sup.groupby(model_col, dropna=False)
            .agg(**agg_dict)
            .reset_index()
            .rename(columns={model_col: "model"})
        )

        colors = (
            sup_conf["model"]
            .astype(str)
            .str.lower()
            .map({
                "lgbm": MODEL_PALETTE["LightGBM"],
                "lightgbm": MODEL_PALETTE["LightGBM"],
                "xgb": MODEL_PALETTE["XGBoost"],
                "xgboost": MODEL_PALETTE["XGBoost"],
            })
            .fillna(NVIDIA_GREEN)
        )

        fig, ax = plt.subplots(figsize=(9, 5))
        ax.bar(sup_conf["model"].astype(str), sup_conf["avg_confidence"], color=colors)
        ax.set_title("Supervised Model Prediction Confidence", fontsize=18)
        ax.set_xlabel("Model")
        ax.set_ylabel("Average Confidence")
        ax.set_ylim(0, 1.05)
        ax.grid(axis="y", alpha=0.15)
        plt.tight_layout()
        save_current_fig("supervised_prediction_confidence")
        plt.show()

else:
    print(f"Missing table: {SUP_FINAL}")


if table_exists(SUP_STATS):
    stats = con.execute(f"SELECT * FROM {SUP_STATS}").df()

    stats_model_col = (
        "model_type" if "model_type" in stats.columns
        else "supervised_model" if "supervised_model" in stats.columns
        else "source_model" if "source_model" in stats.columns
        else None
    )

    metric_candidates = [
        c for c in ["validation_accuracy", "accuracy", "balanced_accuracy", "macro_f1", "weighted_f1"]
        if c in stats.columns
    ]

    if stats_model_col and "stratum" in stats.columns and metric_candidates:
        metric = "weighted_f1" if "weighted_f1" in metric_candidates else metric_candidates[0]

        pivot = stats.pivot_table(
            index="stratum",
            columns=stats_model_col,
            values=metric,
            aggfunc="mean"
        )

        fig, ax = plt.subplots(figsize=(9, 5))
        x = np.arange(len(pivot.index))
        width = 0.8 / max(len(pivot.columns), 1)

        for i, col in enumerate(pivot.columns):
            col_lower = str(col).lower()
            color = (
                MODEL_PALETTE["LightGBM"]
                if "lgb" in col_lower or "light" in col_lower
                else MODEL_PALETTE["XGBoost"]
                if "xgb" in col_lower or "xgboost" in col_lower
                else NVIDIA_GREEN
            )

            offset = (i - (len(pivot.columns) - 1) / 2) * width
            ax.bar(x + offset, pivot[col], width=width, label=str(col), color=color)

        ax.set_xticks(x)
        ax.set_xticklabels([clean_label(s) for s in pivot.index])
        ax.set_title(f"Supervised Model {clean_label(metric)} by Stratum", fontsize=18)
        ax.set_ylabel(clean_label(metric))
        ax.set_ylim(0, 1.05)
        ax.legend()
        ax.grid(axis="y", alpha=0.15)
        plt.tight_layout()
        save_current_fig("supervised_model_validation_metric")
        plt.show()

    else:
        print("Could not build supervised validation plot.")
        print("Available stats columns:", stats.columns.tolist())

else:
    print(f"Missing table: {SUP_STATS}")
```

```python

# ============================================================
# Final visual inventory
# ============================================================

print("Presentation visuals saved in:", PRESENTATION_DIR.resolve())
for p in sorted(PRESENTATION_DIR.glob("*.png")):
    print("-", p.name)

con.close()
print("DuckDB connection closed.")
```
