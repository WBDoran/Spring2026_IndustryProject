```python
import duckdb

con = duckdb.connect("developer_project.duckdb")
```

If you want the same sample every time (important for modeling/debugging)

```python
# Activity sample
con.execute("""
CREATE OR REPLACE TABLE activity_sample AS
SELECT *
FROM activity_clean
ORDER BY RANDOM()
LIMIT 100000
""")
```

```python
# Contact sample
con.execute("""
CREATE OR REPLACE TABLE contact_sample AS
SELECT *
FROM contact_clean
ORDER BY RANDOM()
LIMIT 100000
""")
```

```python
# SDK downloads sample
con.execute("""
CREATE OR REPLACE TABLE sdk_download_sample AS
SELECT *
FROM sdk_download_clean
ORDER BY RANDOM()
LIMIT 100000
""")
```

```python
con.execute("""
COPY activity_sample TO 'Data/activity_sample.csv' (HEADER, DELIMITER ',')
""")

con.execute("""
COPY contact_sample TO 'Data/contact_sample.csv' (HEADER, DELIMITER ',')
""")

con.execute("""
COPY sdk_download_sample TO 'Data/sdk_download_sample.csv' (HEADER, DELIMITER ',')
""")
```
