```python
import duckdb
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

con = duckdb.connect("developer_project.duckdb")

pd.set_option("display.max_columns", 200)
pd.set_option("display.max_rows", 100)
```

```python
# Tables to keep (based on your provided list)
keep_tables = {
    "activity_base_v2",
    "activity_final",
    "activity_ontology_v2",
    "activity_raw",
    "contact_final",
    "contact_one_row_v2",
    "contact_raw",
    "contact_supplement_raw",
    "dev_activation_v2",
    "dev_contact_persona_v2",
    "dev_dormancy_status_v2",
    "dev_features_0_30d_v2",
    "dev_features_30_90d_v2",
    "dev_features_90_180d_v2",
    "dev_features_lifetime_v2",
    "dev_journey_state_v2",
    "dev_meaningful_week_v2",
    "dev_persona_v2",
    "dev_profile_final_v4",
    "dev_recency_features_v2",
    "dev_weekly_features_v2",
    "developer_universe_v2",
    "sdk_download_final",
    "sdk_download_raw"
}

# Get all tables
all_tables = con.execute("SHOW TABLES").fetchdf()["name"].tolist()

# Identify tables to drop
drop_tables = [t for t in all_tables if t not in keep_tables]

# Drop them
for t in drop_tables:
    con.execute(f"DROP TABLE IF EXISTS {t}")

print(f"Dropped {len(drop_tables)} unnecessary tables:")
print(drop_tables)
```

```python
con.close()
```
