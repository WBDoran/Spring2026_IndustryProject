```python
import duckdb
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import sweetviz as sv
```

```python
con = duckdb.connect("developer_project.duckdb")
```

```python
# use con to query the database and load the data into a pandas dataframe
df = con.execute("SELECT * FROM activity_clean ORDER BY random() LIMIT 100000").df()
# use sweetviz to create a report of the dataframe
report = sv.analyze(df)
report.show_html("Activity_Clean_EDA.html")
```

```python
# using con to questy tehe data base for contact_clean tables and load the data into a pandas dataframe
contact_df = con.execute("SELECT * FROM contact_clean ORDER BY random() LIMIT 100000").df()
# use sweetviz to create a report of the dataframe
contact_report = sv.analyze(contact_df)
contact_report.show_html("Contact_Clean_EDA.html")
```

```python
# using con to query the database for the sdk_download_clean table and load the data into a pandas dataframe
sdk_download_df = con.execute("SELECT * FROM sdk_download_clean ORDER BY random() LIMIT 100000").df()
# use sweetviz to create a report of the dataframe
sdk_download_report = sv.analyze(sdk_download_df)
sdk_download_report.show_html("SDK_Download_Clean_EDA.html")
```
